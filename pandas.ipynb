```python
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]

shops = ["편의점", "버스", "영화관", "분식집", "온라인쇼핑", "카페",
         "음원앱", "버스", "편의점", "웹툰", "온라인쇼핑", "분식집"]

categories = ["식비", "교통", "문화", "식비", "쇼핑", "식비",
              "문화", "교통", "식비", "문화", "쇼핑", "식비"]

food = 0
for a, c in zip(amounts, categories):
    if c == "식비":
        food = food + a

print("식비 합계: ", food, "원")
```

```python
import pandas as pd

spend = pd.DataFrame({
    "date": [2, 3, 4, 5, 7, 9, 12, 14, 16, 19, 20, 25],
    "category": categories,
    "shop": shops,
    "amount": amounts
})

spend
```
